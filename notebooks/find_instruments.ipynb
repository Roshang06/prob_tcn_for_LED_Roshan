```python
import pyvisa

rm = pyvisa.ResourceManager()

for resource in rm.list_resources():
    try:
        inst = rm.open_resource(resource)
        inst.timeout = 5000    
        idn = inst.query('*IDN?')
        print(f"{resource}\n    -> {idn.strip()}")
        inst.close()
    except Exception as e:
        print(f"{resource}\n    -> no response / error: {e}")
```

```python
import serial, time
s = serial.Serial('/dev/ttyUSB0', baudrate=9600, timeout=1, rtscts=False)
# RTS toggling should mirror to CTS via the jumper
s.rts = True;  time.sleep(0.05); print("RTS=True  → CTS reads:", s.cts)
s.rts = False; time.sleep(0.05); print("RTS=False → CTS reads:", s.cts)
s.close()

```
